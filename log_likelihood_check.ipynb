```python
from poisson_hypergraph import GH
import xgi
import numpy as np


def generate_graph(true_theta, timesteps):
    true_p, true_q, gamma_nu, gamma_nr, gamma_eu, gamma_er = true_theta
    H = xgi.Hypergraph([[0, 1]])
    H.set_node_attributes({0 : 0, 1 : 1}, name = "label")
    g = GH(H, [0, 1], true_p, true_q)
    g.add_hyperedge(timesteps, gamma_nu, gamma_nr, gamma_eu, gamma_er)
    return g

true_p = .8
true_q = .2
gamma_nu, gamma_nr, gamma_eu, gamma_er = .75, 0.25, 0.75, 0.25
true_theta = [true_p, true_q, gamma_nu, gamma_nr, gamma_eu, gamma_er]
timesteps = 100
g = generate_graph(true_theta, timesteps)




```

```python
# test whether the true labels is a better likelihood estimator compared to a randomly generated label vector
import random

e_index = len(g.get_edges())-1

f_index = 0 # random.randint(0, e_index)

llik_current = g.log_likelihood_given_u_f(f_index, 0, e_index, true_theta, g.get_labels())
llik_my = np.log(g.likelihood_given_u_f(f_index, 0, e_index, true_theta))

differen = llik_my-llik_current

print("difference: " + str(differen))

print("my log likelihood: " + str(g.log_likelihood_given_u_f(f_index, 0, e_index, true_theta, g.get_labels())))

print("current log likelihood: " + str(np.log(g.likelihood_given_u_f(f_index, 0, e_index, true_theta))))







```
