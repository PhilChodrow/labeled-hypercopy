```python
import matplotlib.pyplot as plt
import os

path = os.path.abspath("computermodernstyle.mplstyle")
plt.style.use(path)
```

```python
import csv

def plot_algo_progress(csv_path, title):
    aris = []

    with open(csv_path, 'r', newline="") as f:
        reader = csv.reader(f)

        for row in reader:
            aris.append(float(row[2]))

    plt.title(title)
    plt.plot(range(0, len(aris)), aris)
    plt.xlabel("Timesteps")
    plt.ylabel("ARI")
    plt.show()


plot_algo_progress("../data/senate_max_edge_size15.csv", "senate bills approx edges with intersection size less than 15")
# data/best_senate_bills_attempt.csv
```
